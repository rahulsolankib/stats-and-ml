```python
# import the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# sklearn libraries
from sklearn.preprocessing import OneHotEncoder

# ignore harmless warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
# data import
data = pd.read_csv('/content/drive/MyDrive/Datasets/Input/Housing.csv')

# data sample
data.head()
```

```python
# basic data quality check
pd.DataFrame({'missing_values': data.isnull().sum(),
              'datatype': data.dtypes})
```

# ONE HOT ENCODING

```python
# list of all categorical columns

# list comprehension
cat_cols = [cols for cols in data.columns if data[cols].dtype == 'O']

# classes in each of the columns
for cols in cat_cols:
    print(cols, ':', data[cols].unique())
```

```python
# one hot encoding using sklearn
sk_encoder = OneHotEncoder(sparse_output=False, drop='first', dtype='i2')
out_arr = sk_encoder.fit_transform(data[cat_cols])
sk_encoded_df = pd.DataFrame(out_arr, columns=sk_encoder.get_feature_names_out())

# lets concat the numerical column with encoded columns
sk_encoded_df = pd.concat([data.drop(cat_cols, axis=1), sk_encoded_df], axis=1)
sk_encoded_df.head()
```

```python
# Pandas - get_dummies
get_dummies = pd.get_dummies(data, drop_first=True, dtype='i2')
get_dummies.head()
```
